### **1.Consumo di alcol per paese** 
Esegui un’**analisi esplorativa** focalizzata sul **consumo di alcol**.  
Analizza i seguenti aspetti:
* Visualizza i primi 10 paesi ordinati per total_litres_of_pure_alcohol (dal più alto)

* Calcola la media del consumo di birra, vino, e distillati

* Crea una nuova colonna alcohol_index che sia: `(beer_servings + wine_servings + spirit_servings) / 3`

* Trova il paese con il valore massimo di alcohol_index

* Filtra solo i paesi che consumano più di 100 birre all’anno

Crea un bar chart dei 10 paesi con più consumo totale (total_litres_of_pure_alcohol)

Crea un line plot con wine_servings ordinato per paese (usa sort_values)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("https://raw.githubusercontent.com/fivethirtyeight/data/master/alcohol-consumption/drinks.csv")
top_10_alcohol=df.sort_values(by="total_litres_of_pure_alcohol", ascending=False).head(10)
medie=df[["beer_servings", "wine_servings", "spirit_servings"]].mean()
df["alcohol_index"]=(df["beer_servings"]+df["wine_servings"]+df["spirit_servings"])/3
indice_massimo=df["alcohol_index"].idxmax()
paese_di_ubriachi=df.loc[indice_massimo]
bevitori_birra=df[df["beer_servings"]>100]
plt.figure(figsize=(10, 6))
sns.barplot(data=top_10_alcohol, x="total_litres_of_pure_alcohol", y="country", palette="viridis")
plt.title("Top 10 Paesi per Consumo Totale di Alcol Puro")
plt.xlabel("Litri totali di alcol puro")
plt.ylabel("Paese")
df_ordinato=df.sort_values(by="country")
plt.figure(figsize=(15, 6))
sns.lineplot(data=df_ordinato, x="country", y="wine_servings", marker="o", color="darkred")
plt.title('Andamento del Consumo di Vino per Paese (Ordine Alfabetico)')
plt.xlabel('Paese')
plt.ylabel('Dosi di Vino')
plt.xticks(rotation=90)
print(top_10_alcohol[["country", "total_litres_of_pure_alcohol"]])
print("\n")
print(medie)
print("\n")
print(paese_di_ubriachi["country"])
print("\n")
print(bevitori_birra[["country", "beer_servings"]])
plt.show()
df
```

### **3. Analisi vendite (dataset tips)**

Esegui un’**analisi esplorativa** focalizzata sulle **vendite giornaliere**.  
Analizza i seguenti aspetti per ogni giorno (`day`):

* Il **conto medio** (`total_bill`)
* Il **numero totale di osservazioni** (conteggio delle righe)
* Il **valore minimo e massimo** del conto (`total_bill` → `min` / `max`)

* Crea una nuova colonna `conto_per_persona` che sia:  
  `total_bill / size`

* Trova il giorno con il **conto medio più alto**

Rappresenta graficamente il **conto medio (`total_bill`) per giorno** usando un **grafico a barre**.  
Ordina i dati dal valore più alto al più basso.  
Aggiungi titolo ed etichette agli assi.

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

df["conto_per_persona"]=df["total_bill"]/df["size"]
analisi_giornaliera = df.groupby("day").agg({"total_bill": ["mean", "count", "min", "max"], "conto_per_persona": "mean"})
analisi_giornaliera.columns=["conto medio", "numero righe", "minimo", "massimo", "conto per persona"]
giorno_top=analisi_giornaliera["conto medio"].idxmax()
plot_data = analisi_giornaliera["conto medio"].sort_values(ascending=False)
plt.figure(figsize=(10, 6))
plot_data.plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Conto Medio Giornaliero", fontsize=14)
plt.xlabel("Giorno della Settimana", fontsize=12)
print(analisi_giornaliera)
print("\n")
print(giorno_top)
print("\n")
plt.show()
```
